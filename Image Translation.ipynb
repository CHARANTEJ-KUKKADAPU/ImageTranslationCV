# **Image Translation**

Write a Python program using OpenCV to perform image translation on the **"Given image"**.

The program should:

1. Import `cv2`, `numpy`, and `matplotlib.pyplot`.
2. Read the image **"given** in color mode.
3. Create a **translation matrix** that moves the image  
   **50 pixels to the right** and **80 pixels down**.
4. Apply the translation
5. Display the following using a **1 × 2 layout** in Matplotlib:  
   - **Original Image**  
   - **Translated Image**


```python
# Import 

import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
image = cv2.imread('baseball.jpg')
```

```python

image.shape
```

```python
plt.imshow(image[:,:,::-1])
plt.title('Original Image')
plt.show()
```

```python
tx, ty = 100, 200
M_translation = np.float32([[1, 0, tx], [0, 1, ty]]) 
translated_image = cv2.warpAffine(image, M_translation, (636, 438)) 
```

```python
plt.imshow(translated_image[:,:,::-1])
plt.title("Translated Image")
plt.axis('on')
plt.show()
```

# **Image Scaling**

Write a Python program using OpenCV to perform image scaling on the **given image**.

The program should:

1. Scale the image to:

2. Use `cv2.resize` with appropriate scaling factors.
5. Display the following using a **1 × 3 layout** in Matplotlib:  
   - **Original Image**  
   - **Downscaled Image (0.5×)**

```python
import cv2
import matplotlib.pyplot as plt
```

```python

image = cv2.imread('baseball.jpg')
```

```python
image.shape
```

```python
fx, fy = 2.0, 1.0  
scaled_image = cv2.resize(image, None, fx=fx, fy=fy, interpolation=cv2.INTER_LINEAR)
```

```python
downscaled_image = cv2.resize(image, None, fx=0.5, fy=0.5, interpolation=cv2.INTER_LINEAR)
```

```python
plt.imshow(downscaled_image[:,:,::-1]) 
plt.title("Downscaled Image (0.5x)") 
plt.axis('on')
plt.show()
```

```python
upscaled_image = cv2.resize(image, None, fx=2.0, fy=2.0, interpolation=cv2.INTER_LINEAR)
```

```python
plt.imshow(upscaled_image[:,:,::-1]) 
plt.title("Upscaled Image (2x)") 
plt.axis('on')
plt.show()
```

# **Image Rotation**

Write a Python program using OpenCV to perform **image rotation** on the **given image**.

The program should:

1. Rotate the image by:  
   - **45 degrees**  
   - **90 degrees**
2. Use `cv2.getRotationMatrix2D` and `cv2.warpAffine`.
3. Display the following using a **1 × 3 layout** in Matplotlib:  
   - **Original Image**  
   - **45° Rotated Image**  
   - **90° Rotated Image**


```python
import cv2
import matplotlib.pyplot as plt
```

```python
img = cv2.imread("baseball.jpg")
```

```python
image.shape
```

```python
plt.imshow(image[:,:,::-1])
plt.title('Original Image')
plt.show()
```

```python
(height, width) = image.shape[:2]
angle = 45
center = (width // 2, height // 2)
M_rotation = cv2.getRotationMatrix2D(center, angle, 1)
rotated_image = cv2.warpAffine(image, M_rotation, (width, height))
```

```python
plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(image[:,:,::-1])
plt.title("Original Image")
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(rotated_image[:,:,::-1])
plt.title("45 Degree Rotated Image")
plt.axis('off')

plt.tight_layout()
plt.show()
```

```python
angle = 90
center = (width // 2, height // 2)
M_rotation = cv2.getRotationMatrix2D(center, angle, 1)
rotated_image = cv2.warpAffine(image, M_rotation, (width, height))
```

```python
plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(image[:,:,::-1])
plt.title("Original Image")
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(rotated_image[:,:,::-1])
plt.title("90 Degree Rotated Image")
plt.axis('off')

plt.tight_layout()
plt.show()
```
