# **Image Translation**

Write a Python program using OpenCV to perform image translation on the **"Given image"**.

The program should:

1. Import `cv2`, `numpy`, and `matplotlib.pyplot`.
2. Read the image **"given** in color mode.
3. Create a **translation matrix** that moves the image  
   **50 pixels to the right** and **80 pixels down**.
4. Apply the translation
5. Display the following using a **1 × 2 layout** in Matplotlib:  
   - **Original Image**  
   - **Translated Image**


```python
# Import 

import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
img = cv2.imread("baseball.jpg", cv2.IMREAD_COLOR)
```

```python
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
```

```python
rows, cols = img.shape[:2]
translation_matrix = np.float32([
    [1, 0, 50],
    [0, 1, 80]
])
```

```python

translated_img = cv2.warpAffine(img, translation_matrix, (cols, rows))
translated_rgb = cv2.cvtColor(translated_img, cv2.COLOR_BGR2RGB)
```

```python
plt.figure(figsize=(10, 5))
plt.plot(1, 2, 1)
plt.imshow(img_rgb)
plt.title("Original Image")
plt.axis("off")
```

```python
plt.plot(1, 2, 2)
plt.imshow(translated_rgb)
plt.title("Translated Image")
plt.axis("off")
plt.tight_layout()
plt.show()
```

# **Image Scaling**

Write a Python program using OpenCV to perform image scaling on the **given image**.

The program should:

1. Scale the image to:
   - **0.5 times** its original size (downscaling)
   - **2 times** its original size (upscaling)
2. Use `cv2.resize` with appropriate scaling factors.
5. Display the following using a **1 × 3 layout** in Matplotlib:  
   - **Original Image**  
   - **Downscaled Image (0.5×)**  
   - **Upscaled Image (2×)**


```python
import cv2
import matplotlib.pyplot as plt
```

```python
img = cv2.imread("baseball.jpg", cv2.IMREAD_COLOR)
```

```python
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
```

```python
downscaled = cv2.resize(
    img,
    None,
    fx=0.5,
    fy=0.5,
    interpolation=cv2.INTER_AREA
)
upscaled = cv2.resize(
    img,
    None,
    fx=2,
    fy=2,
    interpolation=cv2.INTER_CUBIC
)
downscaled_rgb = cv2.cvtColor(downscaled, cv2.COLOR_BGR2RGB)
upscaled_rgb = cv2.cvtColor(upscaled, cv2.COLOR_BGR2RGB)
```

```python
plt.figure(figsize=(15, 5))
plt.plot(1, 3, 1)
plt.imshow(img_rgb)
plt.title("Original Image")
plt.axis("off")
```

```python
plt.plot(1, 3, 2)
plt.imshow(downscaled_rgb)
plt.title("Downscaled Image (0.5×)")
plt.axis("off")
```

```python
plt.plot(1, 3, 3)
plt.imshow(upscaled_rgb)
plt.title("Upscaled Image (2×)")
plt.axis("off")
plt.tight_layout()
plt.show()
```

# **Image Shearing**

Write a Python program using OpenCV to perform image shearing on the **given image**.

The program should:

1. Create a **shearing transformation matrix** to apply:
   - **Horizontal shearing**  
   - **Vertical shearing**
2. Apply the shearing transformations.
3. Display the following using a **1 × 3 layout** in Matplotlib:  
   - **Original Image**  
   - **Horizontally Sheared Image**  
   - **Vertically Sheared Image**


```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
img = cv2.imread("baseball.jpg", cv2.IMREAD_COLOR)
```

```python
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
```

```python
rows, cols = img.shape[:2]
horizontal_shear = np.float32([
    [1, 0.5, 0],
    [0, 1, 0]
])
horizontally_sheared = cv2.warpAffine(
    img,
    horizontal_shear,
    (int(cols + rows * 0.5), rows)
)
vertical_shear = np.float32([
    [1, 0, 0],
    [0.5, 1, 0]
])
vertically_sheared = cv2.warpAffine(
    img,
    vertical_shear,
    (cols, int(rows + cols * 0.5))
)
```

```python
horizontal_rgb = cv2.cvtColor(horizontally_sheared, cv2.COLOR_BGR2RGB)
vertical_rgb = cv2.cvtColor(vertically_sheared, cv2.COLOR_BGR2RGB)
```

```python
plt.figure(figsize=(15, 5))
plt.plot(1, 3, 1)
plt.imshow(img_rgb)
plt.title("Original Image")
plt.axis("off")
```

```python
plt.plot(1, 3, 2)
plt.imshow(horizontal_rgb)
plt.title("Horizontally Sheared Image")
plt.axis("off")
```

```python
plt.plot(1, 3, 3)
plt.imshow(vertical_rgb)
plt.title("Vertically Sheared Image")
plt.axis("off")
plt.tight_layout()
plt.show()
```

# **Image Rotation**

Write a Python program using OpenCV to perform **image rotation** on the **given image**.

The program should:

1. Rotate the image by:  
   - **45 degrees**  
   - **90 degrees**
2. Use `cv2.getRotationMatrix2D` and `cv2.warpAffine`.
3. Display the following using a **1 × 3 layout** in Matplotlib:  
   - **Original Image**  
   - **45° Rotated Image**  
   - **90° Rotated Image**


```python
import cv2
import matplotlib.pyplot as plt
```

```python
img = cv2.imread("baseball.jpg")
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
```

```python
rows, cols = img.shape[:2]

center = (cols // 2, rows // 2)

matrix_45 = cv2.getRotationMatrix2D(center, 45, 1)
rotated_45 = cv2.warpAffine(img, matrix_45, (cols, rows))

matrix_90 = cv2.getRotationMatrix2D(center, 90, 1)
rotated_90 = cv2.warpAffine(img, matrix_90, (cols, rows))

rotated_45_rgb = cv2.cvtColor(rotated_45, cv2.COLOR_BGR2RGB)
rotated_90_rgb = cv2.cvtColor(rotated_90, cv2.COLOR_BGR2RGB)
```

```python
plt.figure(figsize=(10,8))
plt.imshow(img_rgb)
plt.title("Original Image")
plt.axis("off")
```

```python
plt.figure(figsize=(10,8))
plt.imshow(rotated_45_rgb)
plt.title("45 Degree Rotated Image")
plt.axis("off")
```

```python
plt.figure(figsize=(10,8))
plt.imshow(rotated_90_rgb)
plt.title("90 Degree Rotated Image")
plt.axis("off")
plt.show()
```
